```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Data/esports_france.csv", sep = ";")

df.head()
```

```python
total_per_year = df.groupby("Year")["Total_Prize"].sum()

plt.figure()
total_per_year.plot()
plt.title("Total Prize Money - French Esports Clubs")
plt.xlabel("Year")
plt.ylabel("Total_Prize")
plt.show()
```
